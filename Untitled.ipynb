```python
import numpy as np

rated = add_injury_rates(df)

# If any column held text like "forty", its type would be "object" instead of a number type
count_cols = ["annual_average_employees", "total_hours_worked",
              "total_dafw_cases", "total_djtr_cases", "total_other_cases"]
print(rated[count_cols].dtypes, "\n")

hours = rated["total_hours_worked"]
employees = rated["annual_average_employees"]
print("Missing hours:", hours.isna().sum())
print("Zero hours:", (hours == 0).sum())
print("Zero employees:", (employees == 0).sum(), "\n")

# A full-time worker logs about 2,000 hours a year, so values far from that suggest typos
hours_per_employee = (hours / employees)[employees > 0]
print("Hours per employee:\n", hours_per_employee.describe(), "\n")

# Dividing by zero hours doesn't raise an error in pandas; it quietly produces inf or NaN
print("TRIR values that aren't real numbers:", (~np.isfinite(rated["trir"])).sum())
print("TRIR (valid values only):\n", rated.loc[np.isfinite(rated["trir"]), "trir"].describe(), "\n")

# The same site submitting more than once
print("Repeated establishment_id rows:", rated["establishment_id"].duplicated().sum())
```

```python
from sitebench.clean import drop_implausible_rows

clean = drop_implausible_rows(df)
rated_clean = add_injury_rates(clean)

dropped = len(df) - len(clean)
print(f"Dropped {dropped:,} of {len(df):,} rows ({dropped / len(df):.1%})")
print(rated_clean["trir"].describe())
```

```python
cols = ["establishment_name", "industry_description", "annual_average_employees",
        "total_hours_worked", "recordable_cases", "trir"]

# The 10 highest rates, to see what these sites actually look like
display(rated_clean.nlargest(10, "trir")[cols])

# How many sites sit above each level? A TRIR of 100 means one injury per full-time worker per year
for cutoff in [50, 100, 200]:
    print(f"TRIR above {cutoff}: {(rated_clean['trir'] > cutoff).sum():,} sites")
```

```python
cases = rated_clean["recordable_cases"]
employees = rated_clean["annual_average_employees"]

for multiplier in [1, 2, 3]:
    n = (cases > employees * multiplier).sum()
    print(f"Cases above {multiplier}x employees: {n:,} sites ({n / len(rated_clean):.2%})")
```
